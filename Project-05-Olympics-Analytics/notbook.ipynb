# 🏅 Phase 1 - Dataset Loading & Exploration

## Objective

Load the Olympics dataset using NumPy, explore its structure, extract individual columns, and calculate total medals for each country.

### Concepts Covered

- np.genfromtxt()
- Header skipping
- String & numeric extraction
- Type conversion
- NumPy array operations

```python
# ==========================================================
# PHASE 1 - DATASET LOADING & EXPLORATION
# ==========================================================

import numpy as np

# ----------------------------------------------------------
# Load Dataset
# ----------------------------------------------------------

data = np.genfromtxt(
    "dataset/olympics.csv",
    delimiter=",",
    skip_header=1,
    dtype=str
)

print("="*80)
print("OLYMPICS DATASET LOADED SUCCESSFULLY")
print("="*80)

# ----------------------------------------------------------
# Dataset Information
# ----------------------------------------------------------

print("\nDataset Shape :", data.shape)

print("\nFirst Five Records")
print("-"*80)
print(data[:5])

# ----------------------------------------------------------
# Extract Columns
# ----------------------------------------------------------

country = data[:,0]

gold = data[:,1].astype(int)

silver = data[:,2].astype(int)

bronze = data[:,3].astype(int)

total_medals = gold + silver + bronze

# ----------------------------------------------------------
# Basic Information
# ----------------------------------------------------------

print("\nNumber of Countries :", len(country))

print("Unique Countries :", len(np.unique(country)))

print("Total Gold Medals :", np.sum(gold))

print("Total Silver Medals :", np.sum(silver))

print("Total Bronze Medals :", np.sum(bronze))

print("Overall Medals :", np.sum(total_medals))

# ----------------------------------------------------------
# Preview
# ----------------------------------------------------------

print("\nFirst Five Countries")
print("-"*80)

for i in range(5):
    print(
        f"{country[i]:15}"
        f" Gold: {gold[i]:2}"
        f" Silver: {silver[i]:2}"
        f" Bronze: {bronze[i]:2}"
        f" Total: {total_medals[i]:3}"
    )

print("\n" + "="*80)
print("PHASE 1 COMPLETED SUCCESSFULLY")
print("="*80)
```

# 🏅 Phase 2, 3 & 4 - Statistical, Performance and Efficiency Analysis

## Objective

Analyze the Olympics dataset using NumPy.

### Topics Covered

- Statistical Analysis
- Country Performance
- Ranking
- Medal Efficiency
- Percentage Analysis
- Leader Identification
- Resourceful NumPy Operations

### NumPy Functions Used

- np.sum()
- np.mean()
- np.max()
- np.min()
- np.argmax()
- np.argmin()
- np.argsort()
- np.unique()


```python
# ==========================================================
# PHASE 2, 3 & 4
# STATISTICS + PERFORMANCE + EFFICIENCY ANALYSIS
# ==========================================================

import numpy as np

print("="*90)
print("              OLYMPICS DATA ANALYSIS REPORT")
print("="*90)

# ==========================================================
# PHASE 2 : STATISTICAL ANALYSIS
# ==========================================================

print("\nPHASE 2 : STATISTICAL ANALYSIS")
print("-"*90)

total_gold = np.sum(gold)
total_silver = np.sum(silver)
total_bronze = np.sum(bronze)
overall_medals = np.sum(total_medals)

avg_gold = np.mean(gold)
avg_silver = np.mean(silver)
avg_bronze = np.mean(bronze)
avg_total = np.mean(total_medals)

max_gold = np.max(gold)
min_gold = np.min(gold)

max_total = np.max(total_medals)
min_total = np.min(total_medals)

print(f"Total Gold Medals      : {total_gold}")
print(f"Total Silver Medals    : {total_silver}")
print(f"Total Bronze Medals    : {total_bronze}")
print(f"Overall Medals         : {overall_medals}")

print()

print(f"Average Gold           : {avg_gold:.2f}")
print(f"Average Silver         : {avg_silver:.2f}")
print(f"Average Bronze         : {avg_bronze:.2f}")
print(f"Average Total Medals   : {avg_total:.2f}")

print()

print(f"Highest Gold           : {max_gold}")
print(f"Lowest Gold            : {min_gold}")

print(f"Highest Total Medals   : {max_total}")
print(f"Lowest Total Medals    : {min_total}")

# ==========================================================
# PHASE 3 : COUNTRY PERFORMANCE
# ==========================================================

print("\n")
print("="*90)
print("PHASE 3 : COUNTRY PERFORMANCE")
print("="*90)

gold_leader = country[np.argmax(gold)]
silver_leader = country[np.argmax(silver)]
bronze_leader = country[np.argmax(bronze)]
overall_leader = country[np.argmax(total_medals)]

print(f"Gold Medal Leader      : {gold_leader}")
print(f"Silver Medal Leader    : {silver_leader}")
print(f"Bronze Medal Leader    : {bronze_leader}")
print(f"Overall Champion       : {overall_leader}")

# ----------------------------------------------------------

top5 = np.argsort(total_medals)[-5:]

print("\nTOP 5 COUNTRIES")
print("-"*90)

print(f"{'Rank':<6}{'Country':<20}{'Gold':<8}{'Silver':<8}{'Bronze':<8}{'Total'}")

for rank, i in enumerate(top5[::-1], start=1):

    print(
        f"{rank:<6}"
        f"{country[i]:<20}"
        f"{gold[i]:<8}"
        f"{silver[i]:<8}"
        f"{bronze[i]:<8}"
        f"{total_medals[i]}"
    )

# ----------------------------------------------------------

bottom5 = np.argsort(total_medals)[:5]

print("\nBOTTOM 5 COUNTRIES")
print("-"*90)

print(f"{'Rank':<6}{'Country':<20}{'Gold':<8}{'Silver':<8}{'Bronze':<8}{'Total'}")

for rank, i in enumerate(bottom5, start=1):

    print(
        f"{rank:<6}"
        f"{country[i]:<20}"
        f"{gold[i]:<8}"
        f"{silver[i]:<8}"
        f"{bronze[i]:<8}"
        f"{total_medals[i]}"
    )

# ==========================================================
# PHASE 4 : MEDAL EFFICIENCY ANALYSIS
# ==========================================================

print("\n")
print("="*90)
print("PHASE 4 : MEDAL EFFICIENCY")
print("="*90)

gold_percent = (gold / total_medals) * 100
silver_percent = (silver / total_medals) * 100
bronze_percent = (bronze / total_medals) * 100

overall_share = (total_medals / overall_medals) * 100

best_gold_eff = np.argmax(gold_percent)
best_silver_eff = np.argmax(silver_percent)
best_bronze_eff = np.argmax(bronze_percent)

print(f"Highest Gold Percentage Country   : {country[best_gold_eff]}")
print(f"Gold Percentage                   : {gold_percent[best_gold_eff]:.2f}%")

print()

print(f"Highest Silver Percentage Country : {country[best_silver_eff]}")
print(f"Silver Percentage                 : {silver_percent[best_silver_eff]:.2f}%")

print()

print(f"Highest Bronze Percentage Country : {country[best_bronze_eff]}")
print(f"Bronze Percentage                 : {bronze_percent[best_bronze_eff]:.2f}%")

# ----------------------------------------------------------

print("\nCOUNTRY MEDAL SHARE")
print("-"*90)

print(f"{'Country':<20}{'Total Medals':<15}{'World Share'}")

for i in range(len(country)):

    print(
        f"{country[i]:<20}"
        f"{total_medals[i]:<15}"
        f"{overall_share[i]:.2f}%"
    )

# ----------------------------------------------------------

print("\n")
print("="*90)
print("ANALYSIS COMPLETED SUCCESSFULLY")
print("="*90)
```

# 🏅 Phase 5 & 6 - Correlation and Ranking Analysis

## Objective

Analyze relationships between medal types and rank countries based on different performance metrics.

### Topics Covered

- Correlation Analysis
- Medal Distribution
- Ranking Analysis
- Best & Worst Performers
- Medal Efficiency

### NumPy Functions Used

- np.corrcoef()
- np.argsort()
- np.argmax()
- np.argmin()
- np.sum()
- np.mean()

```python
# ==========================================================
# PHASE 5 & 6
# CORRELATION + RANKING ANALYSIS
# ==========================================================

print("="*90)
print("          CORRELATION & RANKING ANALYSIS")
print("="*90)

# ==========================================================
# PHASE 5 : CORRELATION ANALYSIS
# ==========================================================

print("\nPHASE 5 : CORRELATION ANALYSIS")
print("-"*90)

corr_gold = np.corrcoef(gold, total_medals)[0,1]
corr_silver = np.corrcoef(silver, total_medals)[0,1]
corr_bronze = np.corrcoef(bronze, total_medals)[0,1]

print(f"Gold vs Total Medals Correlation   : {corr_gold:.3f}")
print(f"Silver vs Total Medals Correlation : {corr_silver:.3f}")
print(f"Bronze vs Total Medals Correlation : {corr_bronze:.3f}")

print("\nINTERPRETATION")
print("-"*90)

if corr_gold > 0.8:
    print("Gold medals have a very strong positive relationship with total medals.")

if corr_silver > 0.8:
    print("Silver medals have a very strong positive relationship with total medals.")

if corr_bronze > 0.8:
    print("Bronze medals have a very strong positive relationship with total medals.")

# ==========================================================
# MEDAL RATIOS
# ==========================================================

print("\nMEDAL RATIOS")
print("-"*90)

gold_ratio = gold / total_medals
silver_ratio = silver / total_medals
bronze_ratio = bronze / total_medals

print(f"{'Country':<20}{'Gold %':<12}{'Silver %':<12}{'Bronze %'}")

for i in range(len(country)):
    print(
        f"{country[i]:<20}"
        f"{gold_ratio[i]*100:<12.2f}"
        f"{silver_ratio[i]*100:<12.2f}"
        f"{bronze_ratio[i]*100:.2f}"
    )

# ==========================================================
# PHASE 6 : RANKING ANALYSIS
# ==========================================================

print("\n")
print("="*90)
print("PHASE 6 : COUNTRY RANKINGS")
print("="*90)

# ----------------------------------------------------------
# TOP GOLD MEDAL COUNTRIES
# ----------------------------------------------------------

top_gold = np.argsort(gold)[-5:]

print("\nTOP 5 GOLD MEDAL COUNTRIES")
print("-"*90)

print(f"{'Rank':<6}{'Country':<20}{'Gold'}")

for rank, i in enumerate(top_gold[::-1], start=1):

    print(
        f"{rank:<6}"
        f"{country[i]:<20}"
        f"{gold[i]}"
    )

# ----------------------------------------------------------
# TOP TOTAL MEDAL COUNTRIES
# ----------------------------------------------------------

top_total = np.argsort(total_medals)[-5:]

print("\nTOP 5 TOTAL MEDAL COUNTRIES")
print("-"*90)

print(f"{'Rank':<6}{'Country':<20}{'Total'}")

for rank, i in enumerate(top_total[::-1], start=1):

    print(
        f"{rank:<6}"
        f"{country[i]:<20}"
        f"{total_medals[i]}"
    )

# ----------------------------------------------------------
# MOST BALANCED COUNTRIES
# ----------------------------------------------------------

balance = np.abs(gold-silver)+np.abs(silver-bronze)

balanced = np.argsort(balance)[:5]

print("\nMOST BALANCED COUNTRIES")
print("-"*90)

print(f"{'Rank':<6}{'Country':<20}{'Balance Score'}")

for rank, i in enumerate(balanced, start=1):

    print(
        f"{rank:<6}"
        f"{country[i]:<20}"
        f"{balance[i]}"
    )

# ----------------------------------------------------------
# LOWEST MEDAL COUNTRIES
# ----------------------------------------------------------

lowest = np.argsort(total_medals)[:5]

print("\nLOWEST MEDAL COUNTRIES")
print("-"*90)

print(f"{'Rank':<6}{'Country':<20}{'Total'}")

for rank, i in enumerate(lowest, start=1):

    print(
        f"{rank:<6}"
        f"{country[i]:<20}"
        f"{total_medals[i]}"
    )

# ----------------------------------------------------------
# BEST COUNTRY SUMMARY
# ----------------------------------------------------------

best = np.argmax(total_medals)

print("\n")
print("="*90)
print("BEST PERFORMING COUNTRY")
print("="*90)

print(f"Country        : {country[best]}")
print(f"Gold           : {gold[best]}")
print(f"Silver         : {silver[best]}")
print(f"Bronze         : {bronze[best]}")
print(f"Total Medals   : {total_medals[best]}")

print("\n")
print("="*90)
print("RANKING ANALYSIS COMPLETED")
print("="*90)
```

Phase 7 - OLYMPICS ANALYTICS DASHBOARD

```python
# ==========================================================
# PHASE 7 - OLYMPICS ANALYTICS DASHBOARD
# ==========================================================

import matplotlib.pyplot as plt
import numpy as np

# ----------------------------------------------------------
# Dashboard Statistics
# ----------------------------------------------------------

total_gold = np.sum(gold)
total_silver = np.sum(silver)
total_bronze = np.sum(bronze)
overall_medals = np.sum(total_medals)

avg_gold = np.mean(gold)
avg_total = np.mean(total_medals)

best_country = country[np.argmax(total_medals)]

# ----------------------------------------------------------
# Top 5 Countries
# ----------------------------------------------------------

top5 = np.argsort(total_medals)[-5:]

# ----------------------------------------------------------
# Country-wise Average Medal Count
# ----------------------------------------------------------

unique_country = np.unique(country)

avg_medals = []

for c in unique_country:
    avg_medals.append(np.mean(total_medals[country == c]))

avg_medals = np.array(avg_medals)

# ----------------------------------------------------------
# Create Dashboard
# ----------------------------------------------------------

fig, ax = plt.subplots(2, 3, figsize=(22, 12))

# ==========================================================
# Chart 1 : Gold Medals by Country
# ==========================================================

ax[0,0].bar(
    country,
    gold,
    color="gold",
    edgecolor="black"
)

ax[0,0].set_title(
    "Gold Medals by Country",
    fontsize=13,
    fontweight="bold"
)

ax[0,0].set_xlabel("Country")
ax[0,0].set_ylabel("Gold Medals")

ax[0,0].tick_params(
    axis="x",
    rotation=90
)

ax[0,0].grid(
    axis="y",
    alpha=0.3
)

# ==========================================================
# Chart 2 : Gold vs Total Medals
# ==========================================================

ax[0,1].scatter(
    gold,
    total_medals,
    s=120,
    color="royalblue",
    edgecolor="black"
)

ax[0,1].set_title(
    "Gold vs Total Medals",
    fontsize=13,
    fontweight="bold"
)

ax[0,1].set_xlabel("Gold Medals")
ax[0,1].set_ylabel("Total Medals")

ax[0,1].grid(alpha=0.3)

# Annotate each point

for i in range(len(country)):
    ax[0,1].text(
        gold[i],
        total_medals[i],
        country[i],
        fontsize=8
    )

# ==========================================================
# Chart 3 : Top 5 Countries by Total Medals
# ==========================================================

ax[0,2].pie(
    total_medals[top5],
    labels=country[top5],
    autopct="%1.1f%%",
    startangle=90
)

ax[0,2].set_title(
    "Top 5 Countries by Total Medals",
    fontsize=13,
    fontweight="bold"
)

# ==========================================================
# Chart 4 : Total Medal Distribution
# ==========================================================

ax[1,0].hist(
    total_medals,
    bins=6,
    color="skyblue",
    edgecolor="black"
)

ax[1,0].set_title(
    "Total Medal Distribution",
    fontsize=13,
    fontweight="bold"
)

ax[1,0].set_xlabel("Total Medals")
ax[1,0].set_ylabel("Number of Countries")

ax[1,0].grid(alpha=0.3)

# ==========================================================
# Chart 5 : Medal Spread
# ==========================================================

box = ax[1,1].boxplot(
    total_medals,
    patch_artist=True
)

box["boxes"][0].set_facecolor("lightgreen")

ax[1,1].set_title(
    "Total Medal Spread",
    fontsize=13,
    fontweight="bold"
)

ax[1,1].set_ylabel("Total Medals")

ax[1,1].grid(alpha=0.3)

# ==========================================================
# Chart 6 : Gold, Silver & Bronze Comparison
# ==========================================================

x = np.arange(len(country))
width = 0.25

ax[1,2].bar(
    x - width,
    gold,
    width,
    label="Gold",
    color="gold",
    edgecolor="black"
)

ax[1,2].bar(
    x,
    silver,
    width,
    label="Silver",
    color="silver",
    edgecolor="black"
)

ax[1,2].bar(
    x + width,
    bronze,
    width,
    label="Bronze",
    color="peru",
    edgecolor="black"
)

ax[1,2].set_title(
    "Gold, Silver & Bronze Comparison",
    fontsize=13,
    fontweight="bold"
)

ax[1,2].set_xlabel("Country")
ax[1,2].set_ylabel("Medals")

ax[1,2].set_xticks(x)
ax[1,2].set_xticklabels(country, rotation=90)

ax[1,2].legend()

ax[1,2].grid(axis="y", alpha=0.3)

# ==========================================================
# Dashboard Title
# ==========================================================

plt.suptitle(
    "🏅 Olympics Analytics Dashboard",
    fontsize=20,
    fontweight="bold"
)

# ==========================================================
# Executive Summary
# ==========================================================

fig.text(
    0.5,
    0.93,
    f"Total Gold: {total_gold} | "
    f"Total Silver: {total_silver} | "
    f"Total Bronze: {total_bronze} | "
    f"Overall Medals: {overall_medals} | "
    f"Average Medals: {avg_total:.2f} | "
    f"Best Country: {best_country}",
    ha="center",
    fontsize=11,
    bbox=dict(
        facecolor="lightyellow",
        edgecolor="gray",
        boxstyle="round,pad=0.5"
    )
)

# ==========================================================
# Footer
# ==========================================================

fig.text(
    0.5,
    0.01,
    "Created by Vaishnavi | NumPy + Matplotlib Olympics Dashboard",
    ha="center",
    fontsize=10,
    color="gray"
)

# ==========================================================
# Adjust Layout
# ==========================================================

plt.tight_layout(rect=[0, 0.03, 1, 0.90])

# ==========================================================
# Save Dashboard
# ==========================================================

fig.savefig(
    "images/olympics_dashboard.png",
    dpi=300,
    bbox_inches="tight"
)

# ==========================================================
# Show Dashboard
# ==========================================================

plt.show()
```
