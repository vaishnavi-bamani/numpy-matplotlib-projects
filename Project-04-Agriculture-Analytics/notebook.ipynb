Phase 1 — Data Understanding (EDA)

Step 1 — Import Libraries

```python
import numpy as np
import matplotlib.pyplot as plt
```

Step 2 — Load Dataset

```python
data = np.genfromtxt(
    "dataset/agriculture.csv",
    delimiter=",",
    skip_header=1,
    dtype=str
)
```

Phase 2 — Extract Columns

```python
district = data[:,0]
crop = data[:,1]
rainfall = data[:,2].astype(float)
temperature = data[:,3].astype(float)
soil = data[:,4].astype(float)
fertilizer = data[:,5].astype(float)
water = data[:,6].astype(float)
yield_ton = data[:,7].astype(float)
production = data[:,8].astype(float)
```

Phase 7 - Corelation Analysis

```python
# ==========================================================
# PHASE 7 - CORRELATION ANALYSIS
# ==========================================================

import numpy as np

# -----------------------------
# Calculate Correlations
# -----------------------------

rainfall_yield = np.corrcoef(rainfall, yield_ton)[0, 1]
fertilizer_production = np.corrcoef(fertilizer, production)[0, 1]
water_yield = np.corrcoef(water, yield_ton)[0, 1]
soil_yield = np.corrcoef(soil, yield_ton)[0, 1]
temperature_yield = np.corrcoef(temperature, yield_ton)[0, 1]

# -----------------------------
# Function to Interpret Correlation
# -----------------------------

def interpret_correlation(r):

    if r >= 0.90:
        return "Very Strong Positive"

    elif r >= 0.70:
        return "Strong Positive"

    elif r >= 0.40:
        return "Moderate Positive"

    elif r >= 0.10:
        return "Weak Positive"

    elif r > -0.10:
        return "No Significant Correlation"

    elif r > -0.40:
        return "Weak Negative"

    elif r > -0.70:
        return "Moderate Negative"

    elif r > -0.90:
        return "Strong Negative"

    else:
        return "Very Strong Negative"

# -----------------------------
# Display Results
# -----------------------------

print("=" * 70)
print("               AGRICULTURE CORRELATION ANALYSIS")
print("=" * 70)

print(f"\nRainfall vs Yield")
print(f"Correlation Coefficient : {rainfall_yield:.3f}")
print(f"Interpretation          : {interpret_correlation(rainfall_yield)}")

print("-" * 70)

print(f"Fertilizer vs Production")
print(f"Correlation Coefficient : {fertilizer_production:.3f}")
print(f"Interpretation          : {interpret_correlation(fertilizer_production)}")

print("-" * 70)

print(f"Water Usage vs Yield")
print(f"Correlation Coefficient : {water_yield:.3f}")
print(f"Interpretation          : {interpret_correlation(water_yield)}")

print("-" * 70)

print(f"Soil Moisture vs Yield")
print(f"Correlation Coefficient : {soil_yield:.3f}")
print(f"Interpretation          : {interpret_correlation(soil_yield)}")

print("-" * 70)

print(f"Temperature vs Yield")
print(f"Correlation Coefficient : {temperature_yield:.3f}")
print(f"Interpretation          : {interpret_correlation(temperature_yield)}")

print("=" * 70)
```

🌾 PHASE 8 — Ranking & Efficiency Analysis

Objective: Rank the best and worst-performing districts and identify the most efficient ones.

```python
# ==========================================================
# PHASE 8 - RANKING & EFFICIENCY ANALYSIS
# ==========================================================

print("=" * 80)
print("            AGRICULTURE RANKING & EFFICIENCY ANALYSIS")
print("=" * 80)

# ----------------------------------------------------------
# Top 5 Districts by Production
# ----------------------------------------------------------

top5_production = np.argsort(production)[-5:]

print("\nTOP 5 DISTRICTS BY PRODUCTION")
print("-" * 80)
print(f"{'Rank':<5}{'District':<15}{'Crop':<15}{'Production'}")
print("-" * 80)

for rank, i in enumerate(top5_production[::-1], start=1):
    print(f"{rank:<5}{district[i]:<15}{crop[i]:<15}{production[i]:.0f} Ton")


# ----------------------------------------------------------
# Bottom 5 Districts by Production
# ----------------------------------------------------------

bottom5_production = np.argsort(production)[:5]

print("\nBOTTOM 5 DISTRICTS BY PRODUCTION")
print("-" * 80)
print(f"{'Rank':<5}{'District':<15}{'Crop':<15}{'Production'}")
print("-" * 80)

for rank, i in enumerate(bottom5_production, start=1):
    print(f"{rank:<5}{district[i]:<15}{crop[i]:<15}{production[i]:.0f} Ton")


# ----------------------------------------------------------
# Top 5 Districts by Yield
# ----------------------------------------------------------

top5_yield = np.argsort(yield_ton)[-5:]

print("\nTOP 5 DISTRICTS BY YIELD")
print("-" * 80)
print(f"{'Rank':<5}{'District':<15}{'Crop':<15}{'Yield'}")
print("-" * 80)

for rank, i in enumerate(top5_yield[::-1], start=1):
    print(f"{rank:<5}{district[i]:<15}{crop[i]:<15}{yield_ton[i]:.2f} Ton")


# ----------------------------------------------------------
# Top 5 Districts by Rainfall
# ----------------------------------------------------------

top5_rainfall = np.argsort(rainfall)[-5:]

print("\nTOP 5 DISTRICTS BY RAINFALL")
print("-" * 80)
print(f"{'Rank':<5}{'District':<15}{'Rainfall'}")
print("-" * 80)

for rank, i in enumerate(top5_rainfall[::-1], start=1):
    print(f"{rank:<5}{district[i]:<15}{rainfall[i]:.0f} mm")


# ----------------------------------------------------------
# Top 5 Districts by Fertilizer Usage
# ----------------------------------------------------------

top5_fertilizer = np.argsort(fertilizer)[-5:]

print("\nTOP 5 DISTRICTS BY FERTILIZER USAGE")
print("-" * 80)
print(f"{'Rank':<5}{'District':<15}{'Fertilizer'}")
print("-" * 80)

for rank, i in enumerate(top5_fertilizer[::-1], start=1):
    print(f"{rank:<5}{district[i]:<15}{fertilizer[i]:.0f} kg")


# ----------------------------------------------------------
# Top 5 Districts by Water Usage
# ----------------------------------------------------------

top5_water = np.argsort(water)[-5:]

print("\nTOP 5 DISTRICTS BY WATER USAGE")
print("-" * 80)
print(f"{'Rank':<5}{'District':<15}{'Water Used'}")
print("-" * 80)

for rank, i in enumerate(top5_water[::-1], start=1):
    print(f"{rank:<5}{district[i]:<15}{water[i]:.0f} L")


# ----------------------------------------------------------
# Water Efficiency
# ----------------------------------------------------------

water_efficiency = production / water

best_water = np.argmax(water_efficiency)

print("\nMOST WATER-EFFICIENT DISTRICT")
print("-" * 50)
print(f"District      : {district[best_water]}")
print(f"Crop          : {crop[best_water]}")
print(f"Production    : {production[best_water]:.0f} Ton")
print(f"Water Used    : {water[best_water]:.0f} L")
print(f"Efficiency    : {water_efficiency[best_water]:.4f} Ton/L")


# ----------------------------------------------------------
# Fertilizer Efficiency
# ----------------------------------------------------------

fertilizer_efficiency = production / fertilizer

best_fertilizer = np.argmax(fertilizer_efficiency)

print("\nMOST FERTILIZER-EFFICIENT DISTRICT")
print("-" * 50)
print(f"District      : {district[best_fertilizer]}")
print(f"Crop          : {crop[best_fertilizer]}")
print(f"Production    : {production[best_fertilizer]:.0f} Ton")
print(f"Fertilizer    : {fertilizer[best_fertilizer]:.0f} kg")
print(f"Efficiency    : {fertilizer_efficiency[best_fertilizer]:.4f} Ton/kg")


print("\n" + "=" * 80)
print("Ranking Analysis Completed Successfully")
print("=" * 80)
```

🌾 PHASE 9 — Executive Agriculture Dashboard

```python
# ==========================================================
# PHASE 9 - AGRICULTURE ANALYTICS DASHBOARD
# ==========================================================

import matplotlib.pyplot as plt
import numpy as np

# ----------------------------------------------------------
# Dashboard Statistics
# ----------------------------------------------------------

total_production = np.sum(production)
avg_yield = np.mean(yield_ton)
avg_rainfall = np.mean(rainfall)
avg_fertilizer = np.mean(fertilizer)
avg_water = np.mean(water)

best_district = district[np.argmax(production)]
best_crop = crop[np.argmax(yield_ton)]

# ----------------------------------------------------------
# Crop-wise Average Production
# ----------------------------------------------------------

unique_crops = np.unique(crop)

crop_avg = []

for c in unique_crops:
    crop_avg.append(np.mean(production[crop == c]))

crop_avg = np.array(crop_avg)

# ----------------------------------------------------------
# Top 5 Water Usage
# ----------------------------------------------------------

top5 = np.argsort(water)[-5:]

# ----------------------------------------------------------
# Create Dashboard
# ----------------------------------------------------------

fig, ax = plt.subplots(2, 3, figsize=(22, 12))

# ==========================================================
# Chart 1 : Production by District
# ==========================================================

ax[0,0].bar(
    district,
    production,
    color="forestgreen",
    edgecolor="black"
)

ax[0,0].set_title("Production by District", fontsize=13)
ax[0,0].set_xlabel("District")
ax[0,0].set_ylabel("Production (Ton)")
ax[0,0].tick_params(axis="x", rotation=90)
ax[0,0].grid(axis="y", alpha=0.3)

# ==========================================================
# Chart 2 : Rainfall vs Yield
# ==========================================================

ax[0,1].scatter(
    rainfall,
    yield_ton,
    color="royalblue",
    edgecolor="black",
    s=80
)

ax[0,1].set_title("Rainfall vs Yield", fontsize=13)
ax[0,1].set_xlabel("Rainfall (mm)")
ax[0,1].set_ylabel("Yield (Ton)")
ax[0,1].grid(alpha=0.3)

# ==========================================================
# Chart 3 : Average Production by Crop
# ==========================================================

ax[0,2].bar(
    unique_crops,
    crop_avg,
    color="orange",
    edgecolor="black"
)

ax[0,2].set_title("Average Production by Crop", fontsize=13)
ax[0,2].set_xlabel("Crop")
ax[0,2].set_ylabel("Average Production")
ax[0,2].tick_params(axis="x", rotation=45)
ax[0,2].grid(axis="y", alpha=0.3)

# ==========================================================
# Chart 4 : Yield Distribution
# ==========================================================

ax[1,0].hist(
    yield_ton,
    bins=8,
    color="gold",
    edgecolor="black"
)

ax[1,0].set_title("Yield Distribution", fontsize=13)
ax[1,0].set_xlabel("Yield (Ton)")
ax[1,0].set_ylabel("Frequency")
ax[1,0].grid(alpha=0.3)

# ==========================================================
# Chart 5 : Production Spread
# ==========================================================

box = ax[1,1].boxplot(
    production,
    patch_artist=True
)

box["boxes"][0].set_facecolor("lightgreen")

ax[1,1].set_title("Production Spread", fontsize=13)
ax[1,1].set_ylabel("Production (Ton)")
ax[1,1].grid(alpha=0.3)

# ==========================================================
# Chart 6 : Top 5 Water Usage
# ==========================================================

ax[1,2].pie(
    water[top5],
    labels=district[top5],
    autopct="%1.1f%%",
    startangle=90
)

ax[1,2].set_title("Top 5 Water Usage")

# ==========================================================
# Main Dashboard Title
# ==========================================================

plt.suptitle(
    "🌾 Maharashtra Agriculture Analytics Dashboard",
    fontsize=20,
    fontweight="bold"
)

# ==========================================================
# Executive Summary
# ==========================================================

fig.text(
    0.5,
    0.93,
    f"Total Production: {total_production:.0f} Ton | "
    f"Average Yield: {avg_yield:.2f} Ton | "
    f"Average Rainfall: {avg_rainfall:.1f} mm | "
    f"Average Fertilizer: {avg_fertilizer:.1f} kg | "
    f"Average Water: {avg_water:.1f} L | "
    f"Best District: {best_district} | "
    f"Best Crop: {best_crop}",
    ha="center",
    fontsize=11,
    bbox=dict(
        facecolor="lightyellow",
        edgecolor="gray",
        boxstyle="round,pad=0.5"
    )
)

# ==========================================================
# Footer
# ==========================================================

fig.text(
    0.5,
    0.01,
    "Created by Vaishnavi | NumPy + Matplotlib Agriculture Dashboard",
    ha="center",
    fontsize=10,
    color="gray"
)

# ==========================================================
# Adjust Layout
# ==========================================================

plt.tight_layout(rect=[0, 0.03, 1, 0.90])

# ==========================================================
# Save Dashboard
# ==========================================================

fig.savefig(
    "images/agriculture_dashboard.png",
    dpi=300,
    bbox_inches="tight"
)

# ==========================================================
# Show Dashboard
# ==========================================================

plt.show()
```
