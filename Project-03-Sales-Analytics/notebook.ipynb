📖 Phase 1 — Load Dataset

```python
import numpy as np
import matplotlib.pyplot as plt
```

Load Dataset

```python
data = np.genfromtxt(
    "dataset/sales.csv",
    delimiter=",",
    skip_header=1
)
```

Extract columns

```python
day = data[:,0]
sales = data[:,1]
print(data)
```

📖 Phase 2 — Statistical Analysis

```python
total_revenue = np.sum(sales)
print(f"Total Revenue: ${total_revenue:.2f}")

Average_daily_revenue = np.mean(sales)
print(f"Average Daily Revenue: ${Average_daily_revenue:.2f}")

highest_revenue_day = day[np.argmax(sales)]
print(f"Highest Revenue Day: Day {highest_revenue_day}")

lowest_revenue_day = day[np.argmin(sales)]
print(f"Lowest Revenue Day: Day {lowest_revenue_day}")

standard_deviation = np.std(sales)
print(f"Standard Deviation of Daily Revenue: ${standard_deviation:.2f}")

variance = np.var(sales)
print(f"Variance of Daily Revenue: ${variance:.2f}")

median_revenue = np.median(sales)
print(f"Median Daily Revenue: ${median_revenue:.2f}")

Highest_sales_days = np.argmax(sales)
print(f"Highest Sales Day: Day {Highest_sales_days}")

lowest_revenue_day = np.argmin(sales)
print(f"Lowest Revenue Day: Day {lowest_revenue_day}")
```

📈 Phase 3 — Trend Analysis

```python
daily_growth = np.diff(sales)
```

Moving Average (3-Day)

```python
moving_average = np.convolve(
    sales,
    np.ones(3)/3,
    mode="valid"
)
```

📊 Phase 4 — Business Insights

```python
days_above_average = day[sales > Average_daily_revenue]
print(f"Days with Revenue Above Average: {days_above_average}")

days_below_average = day[sales < Average_daily_revenue]
print(f"Days with Revenue Below Average: {days_below_average}")

revenue_15_days = sales[:15]
print(f"Revenue for First 15 Days: {revenue_15_days}")

revenue_last_15_days = sales[-15:]
print(f"Revenue for Last 15 Days: {revenue_last_15_days}")

revenue_growth_rate = (sales[-1] - sales[0]) / sales[0] * 100
print(f"Revenue Growth Rate: {revenue_growth_rate:.2f}%")
```

🚀 Phase 5 — Sales Analytics Dashboard

```python
import numpy as np
import matplotlib.pyplot as plt

# ==========================================================
# Statistics
# ==========================================================

total_revenue = np.sum(sales)
average_sale = np.mean(sales)
highest_sale = np.max(sales)
lowest_sale = np.min(sales)

std_sale = np.std(sales)
median_sale = np.median(sales)

days_above = np.sum(sales > average_sale)
days_below = np.sum(sales <= average_sale)

moving_average = np.convolve(
    sales,
    np.ones(3)/3,
    mode="valid"
)

# ==========================================================
# Dashboard
# ==========================================================

fig, ax = plt.subplots(2,3, figsize=(18,10))

# ==========================================================
# 1 Daily Sales Trend
# ==========================================================

ax[0,0].plot(
    day,
    sales,
    color="royalblue",
    marker="o",
    linewidth=2
)

ax[0,0].set_title("Daily Sales Trend")
ax[0,0].set_xlabel("Day")
ax[0,0].set_ylabel("Sales ($)")
ax[0,0].grid(alpha=0.3)

# ==========================================================
# 2 Sales Distribution
# ==========================================================

ax[0,1].hist(
    sales,
    bins=6,
    color="orange",
    edgecolor="black"
)

ax[0,1].set_title("Sales Distribution")
ax[0,1].set_xlabel("Sales ($)")
ax[0,1].set_ylabel("Frequency")
ax[0,1].grid(alpha=0.3)

# ==========================================================
# 3 Sales Summary
# ==========================================================

summary_labels = [
    "Total Revenue",
    "Average Sale",
    "Highest Sale",
    "Lowest Sale"
]

summary_values = [
    total_revenue,
    average_sale,
    highest_sale,
    lowest_sale
]

ax[0,2].bar(
    summary_labels,
    summary_values,
    color=["royalblue","green","gold","red"]
)

ax[0,2].set_title("Sales Summary")
ax[0,2].tick_params(axis="x", rotation=20)
ax[0,2].grid(axis="y", alpha=0.3)

# ==========================================================
# 4 Moving Average
# ==========================================================

ax[1,0].plot(
    day[2:],
    moving_average,
    color="purple",
    linewidth=3
)

ax[1,0].set_title("3-Day Moving Average")
ax[1,0].set_xlabel("Day")
ax[1,0].set_ylabel("Average Sales ($)")
ax[1,0].grid(alpha=0.3)

# ==========================================================
# 5 Box Plot
# ==========================================================

box = ax[1,1].boxplot(
    sales,
    patch_artist=True
)

box["boxes"][0].set_facecolor("lightgreen")

ax[1,1].set_title("Sales Spread")
ax[1,1].set_ylabel("Sales ($)")
ax[1,1].grid(alpha=0.3)

# ==========================================================
# 6 Above vs Below Average
# ==========================================================

ax[1,2].pie(
    [days_above, days_below],
    labels=["Above Avg","Below Avg"],
    autopct="%1.1f%%",
    colors=["limegreen","tomato"],
    explode=(0.05,0),
    startangle=90
)

ax[1,2].set_title("Sales Performance")

# ==========================================================
# Dashboard Title
# ==========================================================

plt.suptitle(
    "📈 Sales Analytics Dashboard",
    fontsize=20,
    fontweight="bold"
)

# ==========================================================
# Summary Statistics
# ==========================================================

fig.text(
    0.5,
    0.93,
    f"Revenue: ${total_revenue:,.0f} | "
    f"Average: ${average_sale:.2f} | "
    f"Highest: ${highest_sale:.0f} | "
    f"Lowest: ${lowest_sale:.0f} | "
    f"Std Dev: {std_sale:.2f}",
    ha="center",
    fontsize=12,
    bbox=dict(
        facecolor="lightyellow",
        edgecolor="gray",
        boxstyle="round,pad=0.5"
    )
)

# ==========================================================
# Footer
# ==========================================================

fig.text(
    0.5,
    0.01,
    "Created by Vaishnavi | NumPy + Matplotlib",
    ha="center",
    fontsize=10,
    color="gray"
)

# ==========================================================
# Layout
# ==========================================================

plt.tight_layout(rect=[0,0.03,1,0.90])

# ==========================================================
# Save Dashboard
# ==========================================================

fig.savefig(
    "images/sales_dashboard.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```
