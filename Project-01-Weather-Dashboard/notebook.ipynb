Import libraries

```python
import numpy as np
import matplotlib.pyplot as plt
```

Step 1 - Load the CSV

```python
data = np.genfromtxt(
    "dataset/weather.csv",
    delimiter=",",
    skip_header=1
)

print(data)
```

Step 2 — Extract Columns

```python
day = data[:,0]

temperature = data[:,1]

rainfall = data[:,2]

humidity = data[:,3]
```

Print shapes

```python
print(day.shape)

print(temperature.shape)

print(rainfall.shape)

print(humidity.shape)
```

Step 3 — Basic Statistics

Calculate:
Average Temperature

Maximum Temperature

Minimum Temperature

Average Rainfall

Maximum Rainfall

Average Humidity

Standard Deviation of Temperature

```python
print("Average temperature:", np.mean(temperature))
print("Maximum temperature:", np.max(temperature))
print("Minimum temperature:", np.min(temperature))
print("Average rainfall:", np.mean(rainfall))
print("Maximum rainfall:", np.max(rainfall))
print("Average humidity:", np.mean(humidity))
print("Standard deviation of humidity:", np.std(humidity))
```

Step 4 — Temperature Trend (Line Plot)

🎯 Goal
Visualize how the temperature changes over the 30 days.

```python
plt.figure(figsize=(10,5))

plt.plot(
    day,
    temperature,
    color="red",
    marker="o",
    linestyle="-",
    linewidth=2,
    markersize=6,
    label="Temperature"
)

plt.title("Daily Temperature Trend")

plt.xlabel("Day")

plt.ylabel("Temperature (°C)")

plt.xticks(day)

plt.grid(alpha=0.3)

plt.legend()

plt.tight_layout()

plt.show()
```

Step 5 — Rainfall Bar Chart

Goal
Show rainfall on each day.

```python
plt.figure(figsize=(10,5))

plt.bar(
    day,
    rainfall,
    color="royalblue",
    edgecolor="black"
)

plt.title("Daily Rainfall")

plt.xlabel("Day")

plt.ylabel("Rainfall (mm)")

plt.xticks(day)

plt.grid(axis="y", alpha=0.3)

plt.tight_layout()

plt.show()
```

Step 6 — Humidity Histogram

Now we'll understand humidity distribution.

```python
plt.figure(figsize=(8,5))

plt.hist(
    humidity,
    bins=8,
    color="seagreen",
    edgecolor="black"
)

plt.title("Humidity Distribution")

plt.xlabel("Humidity (%)")

plt.ylabel("Frequency")

plt.grid(alpha=0.3)

plt.tight_layout()

plt.show()
```

Step 7 — Weather Summary Pie Chart

Let's summarize rainy vs dry days.
Create two values.

```python
rainy_days = np.sum(rainfall > 0)

dry_days = np.sum(rainfall == 0)
```

Now plot.

```python
plt.figure(figsize=(6,6))

plt.pie(
    [rainy_days, dry_days],
    labels=["Rainy Days", "Dry Days"],
    autopct="%1.1f%%",
    colors=["skyblue","gold"],
    startangle=90
)

plt.title("Weather Summary")

plt.show()
```

Step 8 — Build the Weather Dashboard

What is a Dashboard?
A dashboard combines multiple visualizations into one screen.

```python
import matplotlib.pyplot as plt
import numpy as np

fig, ax = plt.subplots(2, 2, figsize=(15, 10))

# -----------------------------
# Temperature Trend
# -----------------------------
ax[0,0].plot(
    day,
    temperature,
    color="red",
    marker="o",
    linewidth=2
)

ax[0,0].set_title("Daily Temperature Trend")
ax[0,0].set_xlabel("Day")
ax[0,0].set_ylabel("Temperature (°C)")
ax[0,0].grid(alpha=0.3)

# -----------------------------
# Rainfall
# -----------------------------
ax[0,1].bar(
    day,
    rainfall,
    color="royalblue",
    edgecolor="black"
)

ax[0,1].set_title("Daily Rainfall")
ax[0,1].set_xlabel("Day")
ax[0,1].set_ylabel("Rainfall (mm)")
ax[0,1].grid(axis="y", alpha=0.3)

# -----------------------------
# Humidity Histogram
# -----------------------------
ax[1,0].hist(
    humidity,
    bins=8,
    color="green",
    edgecolor="black"
)

ax[1,0].set_title("Humidity Distribution")
ax[1,0].set_xlabel("Humidity (%)")
ax[1,0].set_ylabel("Frequency")
ax[1,0].grid(alpha=0.3)

# -----------------------------
# Pie Chart
# -----------------------------
rainy_days = np.sum(rainfall > 0)
dry_days = np.sum(rainfall == 0)

ax[1,1].pie(
    [rainy_days, dry_days],
    labels=["Rainy Days", "Dry Days"],
    autopct="%1.1f%%",
    startangle=90,
    colors=["skyblue", "gold"]
)

ax[1,1].set_title("Weather Summary")

plt.suptitle(
    "Weather Analytics Dashboard",
    fontsize=18,
    fontweight="bold"
)

# ==============================
# Summary Statistics
# ==============================
avg_temp = np.mean(temperature)
max_temp = np.max(temperature)
avg_humidity = np.mean(humidity)

fig.text(
    0.5,
    0.93,
    f"Average Temp: {avg_temp:.1f}°C | Highest Temp: {max_temp:.0f}°C | Average Humidity: {avg_humidity:.1f}%",
    ha="center",
    fontsize=12,
    bbox=dict(facecolor="lightyellow", edgecolor="gray")
)

plt.tight_layout()
fig.savefig(
    "images/weather_dashboard.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()


plt.show()
```
