Step 1

```python
import numpy as np
import matplotlib.pyplot as plt
```

Step 2

```python
data = np.genfromtxt(
    "dataset/student_marks.csv",
    delimiter=",",
    skip_header=1
)

print(data)
```

Step 3

Extract columns.

```python
student = data[:,0]

math = data[:,1]

science = data[:,2]

english = data[:,3]

attendance = data[:,4]
```

Phase 4 — Subject Average Bar Chart
🎯 Goal

Compare the average marks of each subject.

Step 1

Calculate the averages.

```python
avg_math = np.mean(math)
avg_science = np.mean(science)
avg_english = np.mean(english)
```

Step 2

```python
subjects = ["Math", "Science", "English"]

averages = [avg_math, avg_science, avg_english]
```

Step 3

```python
plt.figure(figsize=(8,5))

plt.bar(
    subjects,
    averages,
    color=["royalblue", "seagreen", "orange"],
    edgecolor="black"
)

plt.title("Average Marks by Subject")

plt.xlabel("Subjects")

plt.ylabel("Average Marks")

plt.ylim(0,100)

plt.grid(axis="y", alpha=0.3)

plt.tight_layout()

plt.show()
```

Phase 5 — Attendance vs overall_marks_marks Marks

```python
plt.figure(figsize=(8,5))

plt.scatter(
    attendance,
    overall_marks,
    color="crimson",
    s=100,
    edgecolor="black"
)

plt.title("Attendance vs overall_marks_marks Marks")

plt.xlabel("Attendance (%)")

plt.ylabel("overall_marks_marks Marks")

plt.grid(alpha=0.3)

plt.tight_layout()

plt.show()
```

Phase 6 — overall_marks_marks Marks Distribution

```python
plt.figure(figsize=(8,5))

plt.hist(
    overall_marks,
    bins=6,
    color="gold",
    edgecolor="black"
)

plt.title("Distribution of overall_marks_marks Marks")

plt.xlabel("overall_marks_marks Marks")

plt.ylabel("Number of Students")

plt.grid(alpha=0.3)

plt.tight_layout()

plt.show()
```

Phase 7 — Subject-wise Box Plot

```python
plt.figure(figsize=(8,5))

plt.boxplot(
    [math, science, english],
    tick_labels=["Math", "Science", "English"],
    patch_artist=True
)

plt.title("Marks Distribution by Subject")

plt.ylabel("Marks")

plt.grid(alpha=0.3)

plt.tight_layout()

plt.show()
```

Phase 8 — Pass vs Fail Pie Chart

```python
passed = np.sum(overall_marks >= 40)

failed = np.sum(overall_marks < 40)
```

```python
plt.figure(figsize=(6,6))

plt.pie(
    [passed, failed],
    labels=["Passed", "Failed"],
    autopct="%1.1f%%",
    colors=["limegreen", "tomato"],
    startangle=90
)

plt.title("Pass vs Fail Percentage")

plt.show()
```

Phase 9 — Student Performance Dashboard

```python
import numpy as np
import matplotlib.pyplot as plt

# =====================================================
# Statistics
# =====================================================

avg_math = np.mean(math)
avg_science = np.mean(science)
avg_english = np.mean(english)

passed = np.sum(overall_marks >= 40)
failed = np.sum(overall_marks < 40)

# =====================================================
# Dashboard
# =====================================================

fig, ax = plt.subplots(2, 3, figsize=(18, 10))

# =====================================================
# 1. Subject Average Bar Chart
# =====================================================

subjects = ["Math", "Science", "English"]
averages = [avg_math, avg_science, avg_english]

ax[0,0].bar(
    subjects,
    averages,
    color=["royalblue", "seagreen", "orange"],
    edgecolor="black"
)

ax[0,0].set_title("Average Marks by Subject")
ax[0,0].set_xlabel("Subjects")
ax[0,0].set_ylabel("Average Marks")
ax[0,0].set_ylim(0,100)
ax[0,0].grid(axis="y", alpha=0.3)

# =====================================================
# 2. Attendance vs Overall Marks
# =====================================================

ax[0,1].scatter(
    attendance,
    overall_marks,
    color="crimson",
    edgecolor="black",
    s=90
)

ax[0,1].set_title("Attendance vs Overall Marks")
ax[0,1].set_xlabel("Attendance (%)")
ax[0,1].set_ylabel("Overall Marks")
ax[0,1].grid(alpha=0.3)

# =====================================================
# 3. Overall Marks Histogram
# =====================================================

ax[0,2].hist(
    overall_marks,
    bins=6,
    color="gold",
    edgecolor="black"
)

ax[0,2].set_title("Overall Marks Distribution")
ax[0,2].set_xlabel("Marks")
ax[0,2].set_ylabel("Number of Students")
ax[0,2].grid(alpha=0.3)

# =====================================================
# 4. Subject-wise Box Plot
# =====================================================

box = ax[1,0].boxplot(
    [math, science, english],
    tick_labels=["Math", "Science", "English"],
    patch_artist=True
)

colors = ["royalblue", "seagreen", "orange"]

for patch, color in zip(box["boxes"], colors):
    patch.set_facecolor(color)

ax[1,0].set_title("Subject-wise Marks Distribution")
ax[1,0].set_ylabel("Marks")
ax[1,0].grid(alpha=0.3)

# =====================================================
# 5. Pass vs Fail Pie Chart
# =====================================================

ax[1,1].pie(
    [passed, failed],
    labels=["Passed", "Failed"],
    autopct="%1.1f%%",
    colors=["limegreen", "tomato"],
    startangle=90,
    explode=(0.05,0)
)

ax[1,1].set_title("Pass Percentage")

# =====================================================
# 6. Overall Marks Line Plot
# =====================================================

ax[1,2].plot(
    student,
    overall_marks,
    marker="o",
    color="purple",
    linewidth=2
)

ax[1,2].set_title("Overall Marks by Student")
ax[1,2].set_xlabel("Student ID")
ax[1,2].set_ylabel("Overall Marks")
ax[1,2].grid(alpha=0.3)

# =====================================================
# Dashboard Title
# =====================================================

plt.suptitle(
    "📚 Student Performance Analytics Dashboard",
    fontsize=20,
    fontweight="bold"
)

# =====================================================
# Summary Statistics
# =====================================================

fig.text(
    0.5,
    0.93,
    f"Average Overall: {np.mean(overall_marks):.1f} | "
    f"Highest: {np.max(overall_marks):.1f} | "
    f"Lowest: {np.min(overall_marks):.1f} | "
    f"Pass Rate: {(passed/len(overall_marks))*100:.0f}%",
    ha="center",
    fontsize=12,
    bbox=dict(
        facecolor="lightyellow",
        edgecolor="gray",
        boxstyle="round,pad=0.5"
    )
)

# =====================================================
# Footer
# =====================================================

fig.text(
    0.5,
    0.01,
    "Created by Vaishnavi | NumPy + Matplotlib",
    ha="center",
    fontsize=10,
    color="gray"
)

# =====================================================
# Layout
# =====================================================

plt.tight_layout(rect=[0,0.03,1,0.90])

# =====================================================
# Save Dashboard
# =====================================================

fig.savefig(
    "images/student_dashboard.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```
